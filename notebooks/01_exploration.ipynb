# Exploration Findings (Step 3)

## 1. Dataset overview
- 51,717 rows and 17 columns (Zomato restaurant listings)
- `listed_in(city)` holds 30 Bengaluru areas, not cities (most rows: BTM with 3,279; fewest: New BEL Road with 740)
- Only about 12,500 unique restaurants: 39,218 rows repeat the same name + address, because a restaurant is listed once per category (`listed_in(type)`)

## 2. Missing values (as reported by `info()`)
| Column | Missing | % |
|---|---|---|
| `dish_liked` | 28,078 | 54.3% |
| `rate` | 7,775 | 15.0% |
| `phone` | 1,208 | 2.3% |
| `approx_cost(for two people)` | 346 | 0.7% |
| `rest_type` | 227 | 0.4% |
| `cuisines` | 45 | 0.09% |
| `location` | 21 | 0.04% |

All other columns have no missing values.

## 3. Hidden missing values (look complete but are not)
- `rate`: 2,208 rows contain "NEW" and 69 contain "-"
  - Total unusable ratings: 7,775 + 2,208 + 69 = 10,052 (19.4%)
- `menu_item`: 39,617 rows (76.6%) contain "[]" (an empty list stored as text)
- `reviews_list`: 7,595 rows (14.69%) contain "[]"
- 6,474 rows have both no reviews and no usable rating (85% of the no-review rows), so the link is strong but not complete

## 4. Wrong data types and formats
- `rate` is text: values like "3.9/5", and the same rating appears with and without a space ("3.9/5" vs "3.9 /5")
- `approx_cost(for two people)` is text with commas (e.g., "1,000")
- `votes` is correct (integer)

## 5. Distribution notes
- `votes`: median 41 vs mean 284, so it is heavily skewed by a few very popular restaurants (max 16,832)
- 25% of restaurants have 7 votes or fewer, so a minimum-votes filter is needed when comparing ratings
- Most restaurants cost 300 to 500 for two

## 6. Column decisions
**Drop**
- `url`, `phone`: not useful for analysis
- `dish_liked`: 54% missing
- `menu_item`: 76.6% empty, long text lists
- `reviews_list`: long text blocks, would need text analysis

**Keep**
- `name`, `address`, `location`, `listed_in(city)`
- `online_order`, `book_table`
- `rate`, `votes`
- `rest_type`, `cuisines`
- `approx_cost(for two people)`

## 7. Cleaning plan (Step 4)
1. Rename columns to simpler names
2. Remove duplicates (name + address)
3. Fix `rate`: remove "/5", strip spaces, convert to number ("NEW" and "-" become missing)
4. Fix `approx_cost`: remove commas, convert to number
5. Handle missing values (drop rows with no usable rating; drop or fill the few missing costs)
6. Drop the unneeded columns
7. Save to `data/clean/zomato_clean.csv`
