```python
import pandas as pd

df_raw = pd.read_csv("../data/raw/zomato.csv")
df = df_raw.copy()
print(df.shape)
```

```python
df = df.rename(columns={
    "approx_cost(for two people)": "cost_for_two",
    "listed_in(type)": "listed_type",
    "listed_in(city)": "area",
    "rate": "rating"
})
print(df.columns.tolist())
```

```python
print("Rows before:", len(df))
```

```python
df = df.drop_duplicates(subset=["name", "address"], keep="first")
df = df.reset_index(drop=True)

print("Rows after:", len(df))
```

```python
df.head()
```

```python
df.shape
```

```python
df["rating"] = df["rating"].astype(str).str.replace("/5", "", regex=False).str.strip()
df["rating"] = pd.to_numeric(df["rating"], errors="coerce")
```

```python
print(df["rating"].dtype)
print(df["rating"].isnull().sum())
print(df["rating"].describe())
```

```python
df["cost_for_two"] = df["cost_for_two"].astype(str).str.replace(",", "", regex= False)
df["cost_for_two"] = pd.to_numeric(df["cost_for_two"], errors= "coerce")
```

```python
print(df["cost_for_two"].dtype)
```

```python
print(df["cost_for_two"].isnull().sum())
```

```python
print(df["cost_for_two"].describe())
```

```python
print(df.isnull().sum())
```

```python
print("Rows before:", len(df))
```

```python
df = df.dropna(subset=["cost_for_two", "cuisines", "location", "rest_type"])
df = df.reset_index(drop=True)
```

```python
print("Rows after:", len(df))
```

```python
print(df[["rating", "cost_for_two", "cuisines", "location", "rest_type"]].isnull().sum())
```

```python
cols_to_drop = ["url", "phone", "dish_liked", "menu_item", "reviews_list", "listed_type"]
df = df.drop(columns=cols_to_drop)
```

# Cleaning Summary 

## Starting point
- Raw data: 51,717 rows, 17 columns

## Steps performed
1. **Renamed columns** to simple names: `rate` to `rating`, `approx_cost(for two people)` to `cost_for_two`, `listed_in(city)` to `area` (these are Bengaluru areas, not cities)
2. **Removed duplicates** on name + address: 51,717 to 12,499 rows (one row per restaurant)
3. **Fixed `rating`**: removed "/5" and spaces, converted to numbers; "NEW" and "-" became missing values
4. **Fixed `cost_for_two`**: removed commas, converted to numbers
5. **Handled missing values**:
   - Dropped 123 rows missing cost, cuisines, location, or rest type (12,499 to 12,376)
   - Kept 3,132 restaurants with no rating, because averages in pandas, SQL, and Power BI skip blanks and the rows are still valid for other questions; missing ratings were not filled with an average
6. **Dropped columns**: `url`, `phone`, `dish_liked` (54% missing), `menu_item` (76.6% empty), `reviews_list` (long text), `listed_type` (only the first category per restaurant survived deduplication)

## Result
- Clean data: 12,376 restaurants, 11 columns, saved to `data/clean/zomato_clean.csv`
- 9,244 restaurants have a rating; all other columns are complete

## Notes for analysis
- `votes` and `cost_for_two` are skewed: use the median for a typical value
- Filter by a minimum number of votes when comparing ratings
